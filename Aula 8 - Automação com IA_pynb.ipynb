```
import os
import shutil

# --- Configurações --- #
# Defina o diretório onde os arquivos desorganizados estão localizados.
# Por exemplo: 'C:/Usuarios/SeuUsuario/Downloads' ou '/home/seu_usuario/Downloads'
# CERTIFIQUE-SE DE QUE ESTE DIRETÓRIO ESTÁ CORRETO E CONTÉM OS ARQUIVOS QUE VOCÊ DESEJA ORGANIZAR.
# Uma boa prática é testar em uma cópia dos seus arquivos ou em um diretório temporário.
diretorio_origem = '/content/sample_data' # Exemplo: Usando um diretório de amostra do Colab

# Mapeamento de extensões para nomes de pastas em português
mapeamento_extensoes = {
    'pdf': 'Documentos PDF',
    'docx': 'Documentos Word',
    'doc': 'Documentos Word',
    'xlsx': 'Planilhas Excel',
    'xls': 'Planilhas Excel',
    'pptx': 'Apresentações PowerPoint',
    'ppt': 'Apresentações PowerPoint',
    'jpg': 'Imagens',
    'jpeg': 'Imagens',
    'png': 'Imagens',
    'gif': 'Imagens',
    'bmp': 'Imagens',
    'tiff': 'Imagens',
    'mp3': 'Áudios',
    'wav': 'Áudios',
    'aac': 'Áudios',
    'flac': 'Áudios',
    'mp4': 'Vídeos',
    'avi': 'Vídeos',
    'mov': 'Vídeos',
    'mkv': 'Vídeos',
    'zip': 'Compactados',
    'rar': 'Compactados',
    '7z': 'Compactados',
    'tar': 'Compactados',
    'gz': 'Compactados',
    'txt': 'Textos',
    'log': 'Textos',
    'csv': 'Dados CSV',
    'json': 'Dados JSON',
    'xml': 'Dados XML',
    'py': 'Scripts Python',
    'js': 'Scripts JavaScript',
    'html': 'Web',
    'css': 'Web',
    'php': 'Web',
    'java': 'Código Java',
    'cpp': 'Código C++',
    'c': 'Código C',
    'h': 'Código C/C++', # Para arquivos de cabeçalho C/C++
    'md': 'Documentos Markdown',
    'ipynb': 'Notebooks Colab'
}

# --- Lógica de Organização --- #

# Verifica se o diretório de origem existe
if not os.path.exists(diretorio_origem):
    print(f"Erro: O diretório de origem '{diretorio_origem}' não foi encontrado.")
else:
    print(f"Organizando arquivos em: {diretorio_origem}")

    # Itera sobre todos os itens (arquivos e pastas) no diretório de origem
    for nome_arquivo in os.listdir(diretorio_origem):
        caminho_arquivo = os.path.join(diretorio_origem, nome_arquivo)

        # Verifica se o item é um arquivo (e não um diretório)
        if os.path.isfile(caminho_arquivo):
            # Obtém a extensão do arquivo (ex: '.pdf', '.jpg', '.txt')
            _, extensao_arquivo = os.path.splitext(nome_arquivo)

            # Remove o ponto da extensão e converte para minúsculas
            extensao_limpa = extensao_arquivo[1:].lower()

            # Obtém o nome da pasta em português a partir do mapeamento
            # Se a extensão não estiver no mapeamento, usa 'Outros'
            nome_pasta_final = mapeamento_extensoes.get(extensao_limpa, 'Outros')

            # Cria o caminho completo para a nova pasta de destino
            pasta_destino = os.path.join(diretorio_origem, nome_pasta_final)

            # Se a pasta de destino não existir, cria-a
            if not os.path.exists(pasta_destino):
                os.makedirs(pasta_destino)
                print(f"Pasta '{pasta_destino}' criada.")

            # Cria o caminho completo para o arquivo no destino
            caminho_arquivo_destino = os.path.join(pasta_destino, nome_arquivo)

            # Move o arquivo para a pasta de destino
            try:
                shutil.move(caminho_arquivo, caminho_arquivo_destino)
                print(f"Movido '{nome_arquivo}' para '{pasta_destino}'")
            except Exception as e:
                print(f"Erro ao mover '{nome_arquivo}': {e}")

    print("Organização de arquivos concluída!")
    print("Verifique o diretório de origem para ver as novas pastas e arquivos organizados.")
```

**Missão 2 - Consulta simples de API**

```
import requests

def consultar_cep(cep):
    """
    Consulta a API do ViaCEP para obter informações de endereço a partir de um CEP.
    Inclui tratamento de exceções para requisições e erros da API.
    """
    # Remove caracteres não numéricos do CEP
    cep = ''.join(filter(str.isdigit, cep))

    # Verifica se o CEP tem 8 dígitos
    if len(cep) != 8:
        return {"erro": "O CEP deve conter 8 dígitos numéricos."}

    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        response = requests.get(url)
        response.raise_for_status() # Lança uma exceção para códigos de status HTTP de erro (4xx ou 5xx)

        dados_cep = response.json()

        # Verifica se a API retornou um erro (por exemplo, CEP não encontrado)
        if 'erro' in dados_cep and dados_cep['erro']:
            return {"erro": "CEP não encontrado ou inválido."}
        else:
            return dados_cep

    except requests.exceptions.HTTPError as http_err:
        return {"erro": f"Erro HTTP ao consultar o CEP: {http_err}"}
    except requests.exceptions.ConnectionError as conn_err:
        return {"erro": f"Erro de conexão ao consultar o CEP: {conn_err}. Verifique sua internet."}
    except requests.exceptions.Timeout as timeout_err:
        return {"erro": f"Tempo limite excedido ao consultar o CEP: {timeout_err}."}
    except requests.exceptions.RequestException as req_err:
        return {"erro": f"Erro na requisição ao consultar o CEP: {req_err}"}
    except ValueError:
        return {"erro": "Erro ao decodificar a resposta JSON."}
    except Exception as e:
        return {"erro": f"Ocorreu um erro inesperado: {e}"}

def formatar_endereco(dados):
    """
    Formata os dados de endereço retornados pela API do ViaCEP de maneira legível.
    """
    if "erro" in dados:
        return f"Erro: {dados['erro']}"

    endereco_formatado = f"Endereço: {dados.get('logradouro', 'N/A')}\n"
    endereco_formatado += f"Bairro: {dados.get('bairro', 'N/A')}\n"
    endereco_formatado += f"Cidade: {dados.get('localidade', 'N/A')}\n"
    endereco_formatado += f"Estado: {dados.get('uf', 'N/A')}\n"
    endereco_formatado += f"CEP: {dados.get('cep', 'N/A')}\n"
    endereco_formatado += f"DDD: {dados.get('ddd', 'N/A')}"

    return endereco_formatado

# --- Exemplo de uso ---

# CEP fornecido pelo usuário
cep_usuario = input("Digite o CEP que deseja consultar: ")
print(f"\nConsultando CEP: {cep_usuario}")
dados = consultar_cep(cep_usuario)
print(formatar_endereco(dados))
```

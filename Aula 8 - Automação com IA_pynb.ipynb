**Missão 2 - Consulta simples de API**

```
import requests

def consultar_cep(cep):
    """
    Consulta a API do ViaCEP para obter informações de endereço a partir de um CEP.
    Inclui tratamento de exceções para requisições e erros da API.
    """
    # Remove caracteres não numéricos do CEP
    cep = ''.join(filter(str.isdigit, cep))

    # Verifica se o CEP tem 8 dígitos
    if len(cep) != 8:
        return {"erro": "O CEP deve conter 8 dígitos numéricos."}

    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        response = requests.get(url)
        response.raise_for_status() # Lança uma exceção para códigos de status HTTP de erro (4xx ou 5xx)

        dados_cep = response.json()

        # Verifica se a API retornou um erro (por exemplo, CEP não encontrado)
        if 'erro' in dados_cep and dados_cep['erro']:
            return {"erro": "CEP não encontrado ou inválido."}
        else:
            return dados_cep

    except requests.exceptions.HTTPError as http_err:
        return {"erro": f"Erro HTTP ao consultar o CEP: {http_err}"}
    except requests.exceptions.ConnectionError as conn_err:
        return {"erro": f"Erro de conexão ao consultar o CEP: {conn_err}. Verifique sua internet."}
    except requests.exceptions.Timeout as timeout_err:
        return {"erro": f"Tempo limite excedido ao consultar o CEP: {timeout_err}."}
    except requests.exceptions.RequestException as req_err:
        return {"erro": f"Erro na requisição ao consultar o CEP: {req_err}"}
    except ValueError:
        return {"erro": "Erro ao decodificar a resposta JSON."}
    except Exception as e:
        return {"erro": f"Ocorreu um erro inesperado: {e}"}

def formatar_endereco(dados):
    """
    Formata os dados de endereço retornados pela API do ViaCEP de maneira legível.
    """
    if "erro" in dados:
        return f"Erro: {dados['erro']}"

    endereco_formatado = f"Endereço: {dados.get('logradouro', 'N/A')}\n"
    endereco_formatado += f"Bairro: {dados.get('bairro', 'N/A')}\n"
    endereco_formatado += f"Cidade: {dados.get('localidade', 'N/A')}\n"
    endereco_formatado += f"Estado: {dados.get('uf', 'N/A')}\n"
    endereco_formatado += f"CEP: {dados.get('cep', 'N/A')}\n"
    endereco_formatado += f"DDD: {dados.get('ddd', 'N/A')}"

    return endereco_formatado

# --- Exemplo de uso ---

# CEP fornecido pelo usuário
cep_usuario = input("Digite o CEP que deseja consultar: ")
print(f"\nConsultando CEP: {cep_usuario}")
dados = consultar_cep(cep_usuario)
print(formatar_endereco(dados))
```

```
import time # Importa o módulo time para usar a função sleep para pausas.
import random # Importa o módulo random para gerar temperaturas simuladas aleatoriamente.

# --- Constantes de Configuração --- #
# Define o limite de temperatura. Acima deste valor, um alerta será emitido.
LIMITE_TEMPERATURA = 30
# Define o intervalo em segundos entre cada verificação de temperatura.
INTERVALO_MONITORAMENTO_SEGUNDOS = 5
# Intervalo mínimo e máximo para a temperatura simulada.
TEMP_MIN_SIMULADA = 25.0
TEMP_MAX_SIMULADA = 40.0

def enviar_alerta_email_simulado(mensagem):
    """
    Simula o envio de um e-mail de alerta.
    Em um sistema real, aqui você integraria com um serviço de envio de e-mails (ex: SMTP).
    """
    print("\n--- ALERTA DE E-MAIL SIMULADO ---") # Imprime um cabeçalho para o alerta de e-mail simulado.
    print(f"Para: equipe_monitoramento@empresa.com") # Imprime o destinatário do e-mail simulado.
    print(f"Assunto: ALERTA CRÍTICO: Superaquecimento do Sistema") # Imprime o assunto do e-mail simulado.
    print(f"Mensagem: {mensagem}") # Imprime a mensagem principal do e-mail simulado.
    print("----------------------------------") # Imprime um rodapé para o alerta de e-mail simulado.

# --- Programa Autônomo de Monitoramento --- #

print("Iniciando monitoramento autônomo do sistema...") # Informa que o monitoramento está começando.
print(f"Limite de temperatura para alerta: {LIMITE_TEMPERATURA}°C") # Exibe o limite de temperatura configurado.
print(f"Intervalo de verificação: {INTERVALO_MONITORAMENTO_SEGUNDOS} segundos") # Exibe o intervalo de verificação configurado.
print("Pressione 'Interromper execução' (quadrado no Colab) para parar o monitoramento.") # Orienta o usuário sobre como parar o script.

while True: # Inicia um loop infinito para monitoramento contínuo.
    # Simula a temperatura atual do sistema (gerando um valor aleatório para autonomia)
    # Em um sistema real, você leria de um sensor: temperatura_atual = ler_sensor_temperatura()
    temperatura_atual = round(random.uniform(TEMP_MIN_SIMULADA, TEMP_MAX_SIMULADA), 1) # Gera uma temperatura aleatória e arredonda para uma casa decimal.

    print(f"\nTemperatura atual do sistema: {temperatura_atual}°C") # Exibe a temperatura atual simulada.

    # Verifica se a temperatura atual ultrapassa o limite definido.
    if temperatura_atual > LIMITE_TEMPERATURA: # Verifica se a temperatura excede o limite.
        mensagem_alerta = ( # Cria a mensagem de alerta formatada.
            f"ALERTA: Superaquecimento detectado! "
            f"Temperatura ({temperatura_atual}°C) acima de {LIMITE_TEMPERATURA}°C."
        )
        print(mensagem_alerta) # Imprime a mensagem de alerta no console.
        enviar_alerta_email_simulado(mensagem_alerta) # Chama a função para simular o envio de um e-mail de alerta.
    else: # Caso a temperatura esteja normal.
        print("Temperatura do sistema está normal.") # Informa que a temperatura está dentro do limite.

    # Aguarda um tempo antes da próxima verificação para simular monitoramento contínuo
    time.sleep(INTERVALO_MONITORAMENTO_SEGUNDOS) # Pausa a execução pelo tempo definido antes da próxima iteração do loop.
```

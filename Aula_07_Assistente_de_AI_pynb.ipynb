Calculadora simples

```python
# Função para realizar adição
def adicionar(x, y):
    return x + y

# Função para realizar subtração
def subtrair(x, y):
    return x - y

# Loop principal para a calculadora
while True:
    # Solicita ao usuário a expressão ou para sair
    expressao = input("Digite a expressão (ex: 5 + 3 ou 8 - 2), ou 'sair' para encerrar: ")

    # Verifica se o usuário quer sair
    if expressao.lower() == "sair":
        print("Calculadora encerrada.")
        break

    # Divide a expressão em partes (número, operador, número)
    partes = expressao.split()

    # Valida o formato da expressão
    if len(partes) != 3:
        print("Formato de expressão inválido. Use 'número operador número'.")
        continue

    try:
        # Converte o primeiro número para float
        num1 = float(partes[0])
        # Obtém o operador
        operador = partes[1]
        # Converte o segundo número para float
        num2 = float(partes[2])

        # Realiza a operação com base no operador
        if operador == '+':
            # Executa a adição e imprime o resultado
            print(f"{num1} + {num2} = {adicionar(num1, num2)}")
        elif operador == '-':
            # Executa a subtração e imprime o resultado
            print(f"{num1} - {num2} = {subtrair(num1, num2)}")
        else:
            # Informa sobre operador inválido
            print("Operador inválido. Use '+' para adição ou '-' para subtração.")
    except ValueError:
        # Captura erro se os números não forem válidos
        print("Entrada inválida. Certifique-se de que os números sejam válidos.")
    except Exception as e:
        # Captura outros erros inesperados
        print(f"Ocorreu um erro: {e}")

    # Pergunta se o usuário deseja fazer outro cálculo
    proximo_calculo = input("Deseja fazer outro cálculo? (sim/não): ")
    # Se a resposta for 'não', encerra o loop
    if proximo_calculo.lower() == "não":
        print("Calculadora encerrada.")
        break
```
